```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
netflix = pd.read_csv('/netflix_titles.csv')

netflix.head()
# dataset information
netflix.info()
# check missing values
netflix.isnull().sum()
# shape of dataset
netflix.shape# dataset information
netflix.info()
# fill missing values
netflix['director'] = netflix['director'].fillna('Unknown')
netflix['cast'] = netflix['cast'].fillna('Unknown')
netflix['country'] = netflix['country'].fillna('Unknown')
netflix['rating'] = netflix['rating'].fillna('Not Rated')

# remove missing dates
netflix.dropna(subset=['date_added'], inplace=True)
plt.figure(figsize=(6,4))
sns.countplot(x='type', data=netflix)
plt.title('Movies vs TV Shows on Netflix')
plt.show()
plt.figure(figsize=(12,6))
netflix['country'].value_counts().head(10).plot(kind='bar')
plt.title('Top 10 Countries Producing Netflix Content')
plt.xlabel('Country')
plt.ylabel('Count')
plt.show()
plt.figure(figsize=(10,5))
sns.countplot(y='rating', data=netflix, order=netflix['rating'].value_counts().index)
plt.title('Distribution of Content Ratings')
plt.show()
# convert date_added column to datetime
netflix['date_added'] = pd.to_datetime(netflix['date_added'], errors='coerce')

# create year_added column
netflix['year_added'] = netflix['date_added'].dt.year

# plot graph
plt.figure(figsize=(12,6))
netflix['year_added'].value_counts().sort_index().plot(kind='line', marker='o')

plt.title('Netflix Content Added Over the Years')
plt.xlabel('Year')
plt.ylabel('Number of Shows Added')

plt.show()
plt.figure(figsize=(12,6))
netflix['listed_in'].value_counts().head(10).plot(kind='barh')
plt.title('Top 10 Genres on Netflix')
plt.xlabel('Count')
plt.show()
```
